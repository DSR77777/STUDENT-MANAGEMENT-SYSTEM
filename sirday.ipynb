```python
#Array dimensions and shape
a=np.array([[1,2,3],[4,5,6]])
print(a.ndim) #no of axes
print(a.shape) #rows and colums
print(a.size) #size
print(a.dtype) #daratype
```

```python
#Numpy operations
a=np.array([1,2,3])
b=np.array([4,5,6])

print(a+b) #addition
print(a-b) #substraction
print(a**b) #multiplication
print(np.sqrt(a)) #sqroot
print(np.dot(a,b)) #1*4 + 2*5 +3*6=32 
```

```python
a=np.array([3,1,4,1,5,9,2,6])
print(np.sum(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.min(a))
print(np.max(a))
print(np.argmax(a))

m=np.array([[1,2,3],[4,5,6]])
print(np.sum(m,axis=0)) #col sum
print(np.sum(m,axis=1)) #row sum
```
